### 68 Cerita (https://budaya.jogjaprov.go.id/attachment/view?id=27234&&filename=Buku%20Dongeng%202023_compressed.pdf)

```python
import pdfplumber

text = ""

with pdfplumber.open("U1gbOp-Buku Dongeng 2023_compressed.pdf") as pdf:
    for page in pdf.pages:
        text += page.extract_text() + "\n"
```

```python
stories = text.split("Oleh :")
```

```python
import re

cleaned_stories = []

for story in stories[1:]:
    story = story.strip()
    
    story = re.sub(r'\n+', '\n', story)
    
    lines = story.split("\n")
    
    story_clean = "\n".join(lines[1:])
    
    cleaned_stories.append(story_clean)
```

```python
final_stories = [s for s in cleaned_stories if len(s) > 200]
```

```python
import json

data = []

for i, story in enumerate(final_stories):
    data.append({
        "isi": story
    })

with open("dataset_cerita.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2)
```

### Dataset HTML (https://github.com/trionoputra/cerita-rakyat)

```python
import os
import json
import re
from bs4 import BeautifulSoup

folder_path = r"C:\Semester 4\NLP\TUGAS KELOMPOK\data\data_github"
html_files = []

if not os.path.exists(folder_path):
    print(f"Error: Folder '{folder_path}' tidak ditemukan.")
else:
    html_files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.endswith('.html')]

new_data = []

# TAHAP 1: Ekstraksi HTML ke Dataset Terpisah
print("Memulai ekstraksi HTML...")
for file_path in html_files:
    try:
        with open(file_path, "r", encoding="utf-8") as f:
            soup = BeautifulSoup(f, "html.parser")
            paragraphs = soup.find_all('p')
            
            if len(paragraphs) == 0:
                continue

            # Ambil teks dari setiap paragraf
            text_lines = [p.get_text(strip=True) for p in paragraphs]
            
            # Buang judul (elemen pertama) jika teksnya cukup pendek (kurang dari 100 karakter)
            start_index = 0
            if len(text_lines) > 1 and len(text_lines[0]) < 100:
                start_index = 1
            
            # Gabungkan sisa teks menjadi satu kesatuan
            story_text = "\n".join(text_lines[start_index:])
            
            # LOGIKA REGEX PEMBUNUH TAG DAERAH
            # Menghancurkan pola "( Cerita Rakyat ... )" beserta enter/spasi di sekitarnya
            story_text = re.sub(r'^\s*\([\s\S]*?Cerita\s+Rakyat[\s\S]*?\)\s*', '', story_text, flags=re.IGNORECASE)
            
            # Bersihkan enter yang berlebihan akibat penggabungan
            story_text = re.sub(r'\n+', '\n', story_text).strip()
            
            # Filter teks pendek (pastikan itu benar-benar cerita)
            if len(story_text) > 200:
                new_data.append({"isi": story_text})
                
    except Exception as e:
        print(f"Gagal memproses {file_path}: {e}")

# Simpan hasil HTML ke file JSON tersendiri
with open("dataset_html.json", "w", encoding="utf-8") as f:
    json.dump(new_data, f, ensure_ascii=False, indent=2)
    
print(f"Berhasil mengekstrak {len(new_data)} cerita dari HTML ke 'dataset_html.json'.")
```

### Dataset Kalarahu (https://repositori.kemendikdasmen.go.id/2849/1/Kalarahu%20Kumpulan%20Cerita%20Rakyat%20Jawa.pdf)

```python
import pdfplumber
import re
import json

pdf_path = r"C:\Semester 4\NLP\Pembelajaran Bahasa Mendalam\Tugas NER\raw data\Kalarahu Kumpulan Cerita Rakyat Jawa.pdf"
full_text = ""

# 1. Ekstrak teks dari halaman 8 (indeks 7) sampai 64 (indeks 63)
with pdfplumber.open(pdf_path) as pdf:
    for i in range(7, 64):
        page = pdf.pages[i]
        text = page.extract_text()
        if text:
            full_text += text + "\n"

# 2. Definisikan Regex yang sudah disesuaikan dengan "typo" hasil bacaan PDF
titles_regex = [
    r"1\.\s*KALARAHU",
    r"2\.\s*JAKA\s*TARUB\s*DAN\s*NAWANGWULAN",
    r"3\.\s*ASAL\s*MULA\s*NAMA\s*BANYUWANGI",
    r"4\.\s*BAWANG\s*PUTIH\.?\s*DAN\s*BAWANG\s*MERAH",  # Menangani typo "PUTIH."
    r"5\.\s*ASAL\s*MULA\s*HURUF\s*JAWA",
    r"6\.\s*(?:SI|51)\s*WURAGIL",                       # Menangani typo "51 WURAGIL"
    r"7\.\s*BANDUNG\s*BANDAWASA\s*MENGUTUK[\s\n]*LORO\s*JONGGRANG\s*MENJADI\s*ARCA",
    r"8\.\s*DEWI\s*SRI\s*DAN\s*SEDANA",
    r"9\.\s*ANDE[\-\s]*ANDE\s*LUMUT\s*DAN\s*KLETING\s*KUNING",
    r"10\.\s*LARANGAN\s*MEMUKUL\s*KENTONGAN"
]

# Gabungkan dengan OR (|)
pattern = r"(" + "|".join(titles_regex) + r")"

# 3. Potong teks menggunakan pola tersebut (Tambahkan re.IGNORECASE)
parts = re.split(pattern, full_text, flags=re.IGNORECASE)

stories = []

# parts hasil split akan berisi: 
# [0] teks sisa sebelum judul 1
# [1] judul 1 (hasil match)
# [2] isi cerita 1
# ... dst. Kita loop lompat 2.
for i in range(2, len(parts), 2):
    story_text = parts[i]
    
    # 4. Cleaning hardcore agar JSON bersih
    story_text = re.sub(r"\n\d+\n", " ", story_text)           # Menghapus nomor halaman
    story_text = re.sub(r"-\n", "", story_text)                # Menyambung kata yang terpotong strip di ujung baris
    story_text = re.sub(r"\n", " ", story_text)                # Menghapus sisa karakter enter
    story_text = re.sub(r"\s+", " ", story_text)               # Membersihkan spasi ganda
    story_text = re.sub(r"[^\w\s.,\"'()-]", "", story_text)    # Membuang karakter aneh/simbol
    story_text = story_text.strip()
    
    if story_text:
        stories.append({
            "isi": story_text
        })

# 5. Simpan ke JSON
output_path = "kalarahu_clean_final.json"
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(stories, f, ensure_ascii=False, indent=2)

print(f"Proses selesai. Berhasil mengekstrak {len(stories)} cerita!")
```

### Dataset Nusantara (https://dn790000.ca.archive.org/0/items/CeritaAnak/Cerita%20Rakyat%20Nusantara%202.pdf)

```python
import pdfplumber
import json
import re

file_pdf = r"C:\Semester 4\NLP\Pembelajaran Bahasa Mendalam\Tugas NER\raw data\Cerita Rakyat Nusantara 2.pdf"

daftar_judul = [
    "PUTRI KANDITA",
    "LEGENDA GUNUNG ARJUNA",
    "KISAH DOYAN NADA",
    "PANGERAN PANDE GELANG DAN PUTRI CADASARI",
    "PEU MANA MEINEGAKA SAWAI",
    "KISAH DI GUA KISKENDA",
    "KI AGENG PANDANARAN"
]

judul_dilewati = "PANGERAN PANDE GELANG DAN PUTRI CADASARI"
new_data = []

with pdfplumber.open(file_pdf) as pdf:
    full_text = ""
    for page in pdf.pages:
        text = page.extract_text()
        if text:
            full_text += text + "\n"

# ── Bersihkan noise PDF ──────────────────────────────────────────────────────
full_text = re.sub(r'Sumber\s*:\s*http://ceritarakyatnusantara\.com', '', full_text, flags=re.IGNORECASE)
full_text = re.sub(r'CERITA RAKYAT NUSANTARA 2', '', full_text)
full_text = re.sub(r'Cerita Rakyat Nusantara', '', full_text)
full_text = re.sub(r'Merajut Kearifan Menjemput Zaman', '', full_text, flags=re.IGNORECASE)
full_text = re.sub(r'^\s*\d+\s*$', '', full_text, flags=re.MULTILINE)          # nomor halaman
full_text = re.sub(r'^No\s+Judul Cerita\s+Hal\s*$', '', full_text, flags=re.MULTILINE)  # header daftar isi
full_text = re.sub(r'^\d+\s+\S.*?\s+\d+\s*$', '', full_text, flags=re.MULTILINE)       # baris daftar isi

# ── Split berdasarkan heading ALL-CAPS di baris tersendiri ───────────────────
# Strategi ini memastikan hanya judul heading yang menjadi pemotong,
# bukan nama tokoh yang muncul di dalam paragraf cerita.
pola_heading = r'(?:^|\n)(' + '|'.join([re.escape(j) for j in daftar_judul]) + r')(?:\n)'
potongan = re.split(pola_heading, full_text)

for i in range(1, len(potongan) - 1, 2):
    judul_ditemukan = potongan[i].strip()
    isi_cerita = potongan[i + 1]

    # Lewati cerita yang dikecualikan
    if judul_ditemukan.upper() == judul_dilewati.upper():
        continue

    # Hapus baris "Asal cerita : ..."
    isi_cerita = re.sub(r'^Asal cerita\s*:.*?\n', '', isi_cerita, flags=re.IGNORECASE)

    # Potong segala teks sebelum kata "Alkisah" (caption foto, dll.)
    isi_cerita = re.sub(r'^.*?(?=Alkisah)', '', isi_cerita, flags=re.DOTALL)

    # Sambung kata yang terpotong tanda hubung di ujung baris
    isi_cerita = re.sub(r'-\n', '', isi_cerita)

    # Gabungkan baris dalam satu paragraf (enter tunggal → spasi)
    isi_cerita = re.sub(r'(?<!\n)\n(?!\n)', ' ', isi_cerita)

    # Bersihkan spasi ganda
    isi_cerita = re.sub(r'[ \t]+', ' ', isi_cerita)

    # Kurangi baris kosong berlebih
    isi_cerita = re.sub(r'\n{3,}', '\n\n', isi_cerita)

    isi_cerita = isi_cerita.strip()

    if len(isi_cerita) > 200:
        new_data.append({"isi": isi_cerita})

# ── Simpan ke JSON ───────────────────────────────────────────────────────────
output_path = "dataset_nusantara2.json"
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(new_data, f, ensure_ascii=False, indent=2)

print(f"Eksekusi Selesai! Berhasil mengekstrak {len(new_data)} cerita.")
```
